```python
import os
import pandas as pd
import numpy as np

dataset_path = '/Users/simrahruq/Desktop/CIC IoT-DIAD 2024 Raw'  # adjust to your actual local path
print(os.listdir(dataset_path))
```

```python
dfs = []
column_sets = {}

for class_folder in os.listdir(dataset_path):
    class_path = os.path.join(dataset_path, class_folder)
    if not os.path.isdir(class_path):
        continue

    for item in os.listdir(class_path):
        item_path = os.path.join(class_path, item)

        if item.endswith('.csv'):
            df = pd.read_csv(item_path, nrows=5)  # just peek, don't load fully yet
            column_sets[f"{class_folder}/{item}"] = tuple(df.columns)

        elif os.path.isdir(item_path):
            for file_name in os.listdir(item_path):
                if file_name.endswith('.csv'):
                    file_path = os.path.join(item_path, file_name)
                    df = pd.read_csv(file_path, nrows=5)
                    column_sets[f"{class_folder}/{item}/{file_name}"] = tuple(df.columns)

# Find how many unique column structures exist
unique_structures = set(column_sets.values())
print(f"Number of unique column structures: {len(unique_structures)}")
print(f"Number of columns in each structure:")
for structure in unique_structures:
    matching_files = [f for f, cols in column_sets.items() if cols == structure]
    print(f"\n  {len(structure)} columns — used by {len(matching_files)} files")
    print(f"  Example file: {matching_files[0]}")
```

```python
# Get one example file from each unique structure
examples = {}
seen_structures = set()

for file_path, cols in column_sets.items():
    if cols not in seen_structures:
        seen_structures.add(cols)
        examples[file_path] = cols

# Print column names side by side for comparison
for file_path, cols in examples.items():
    print(f"\n{file_path}")
    print(list(cols))
```

```python
# The correct, canonical column names (from a properly-headered file)
correct_columns = ['Flow ID', 'Src IP', 'Src Port', 'Dst IP', 'Dst Port', 'Protocol', 'Timestamp',
    'Flow Duration', 'Total Fwd Packet', 'Total Bwd packets', 'Total Length of Fwd Packet',
    'Total Length of Bwd Packet', 'Fwd Packet Length Max', 'Fwd Packet Length Min',
    'Fwd Packet Length Mean', 'Fwd Packet Length Std', 'Bwd Packet Length Max',
    'Bwd Packet Length Min', 'Bwd Packet Length Mean', 'Bwd Packet Length Std',
    'Flow Bytes/s', 'Flow Packets/s', 'Flow IAT Mean', 'Flow IAT Std', 'Flow IAT Max',
    'Flow IAT Min', 'Fwd IAT Total', 'Fwd IAT Mean', 'Fwd IAT Std', 'Fwd IAT Max', 'Fwd IAT Min',
    'Bwd IAT Total', 'Bwd IAT Mean', 'Bwd IAT Std', 'Bwd IAT Max', 'Bwd IAT Min',
    'Fwd PSH Flags', 'Bwd PSH Flags', 'Fwd URG Flags', 'Bwd URG Flags', 'Fwd Header Length',
    'Bwd Header Length', 'Fwd Packets/s', 'Bwd Packets/s', 'Packet Length Min',
    'Packet Length Max', 'Packet Length Mean', 'Packet Length Std', 'Packet Length Variance',
    'FIN Flag Count', 'SYN Flag Count', 'RST Flag Count', 'PSH Flag Count', 'ACK Flag Count',
    'URG Flag Count', 'CWR Flag Count', 'ECE Flag Count', 'Down/Up Ratio',
    'Average Packet Size', 'Fwd Segment Size Avg', 'Bwd Segment Size Avg',
    'Fwd Bytes/Bulk Avg', 'Fwd Packet/Bulk Avg', 'Fwd Bulk Rate Avg', 'Bwd Bytes/Bulk Avg',
    'Bwd Packet/Bulk Avg', 'Bwd Bulk Rate Avg', 'Subflow Fwd Packets', 'Subflow Fwd Bytes',
    'Subflow Bwd Packets', 'Subflow Bwd Bytes', 'FWD Init Win Bytes', 'Bwd Init Win Bytes',
    'Fwd Act Data Pkts', 'Fwd Seg Size Min', 'Active Mean', 'Active Std', 'Active Max',
    'Active Min', 'Idle Mean', 'Idle Std', 'Idle Max', 'Idle Min', 'Label']

print(f"Number of correct columns: {len(correct_columns)}")  # should be 84
```

```python
dfs = []

for class_folder in os.listdir(dataset_path):
    class_path = os.path.join(dataset_path, class_folder)
    if not os.path.isdir(class_path):
        continue

    print(f"\n--- Starting {class_folder} ---")

    def load_csv_safe(file_path, class_folder):
        # Peek at first row to check if it looks like a header or data
        first_row = pd.read_csv(file_path, nrows=1)
        first_val = str(first_row.columns[0])
        
        # Headerless files have an IP-flow-ID-like string as the "column name"
        looks_like_data = '-' in first_val and any(c.isdigit() for c in first_val)
        
        if looks_like_data:
            # No real header — read again, force our correct column names
            df = pd.read_csv(file_path, header=None, names=correct_columns)
        else:
            # Has a proper header
            df = pd.read_csv(file_path)
        
        df['Label'] = class_folder
        return df

    for item in os.listdir(class_path):
        item_path = os.path.join(class_path, item)

        if item.endswith('.csv'):
            df = load_csv_safe(item_path, class_folder)
            dfs.append(df)
            print(f"  ✓ {item} — {len(df):,} rows")

        elif os.path.isdir(item_path):
            for file_name in os.listdir(item_path):
                if file_name.endswith('.csv'):
                    file_path = os.path.join(item_path, file_name)
                    df = load_csv_safe(file_path, class_folder)
                    dfs.append(df)
                    print(f"  ✓ {item}/{file_name} — {len(df):,} rows")

combined_df = pd.concat(dfs, ignore_index=True)
print(f"\nTotal shape: {combined_df.shape}")
print(combined_df['Label'].value_counts())
```

```python
import psutil

print(f"Available RAM: {psutil.virtual_memory().available / 1e9:.1f} GB")
print(f"Dataset memory usage: {combined_df.memory_usage(deep=True).sum() / 1e9:.2f} GB")
```

```python
SAMPLES_PER_CLASS = {
    'Benign': 398330,
    'DDoS': 300000,
    'DoS': 300000,
    'Mirai': 174588,
    'Spoofing': 157238
}

sampled_dfs = []
for label in combined_df['Label'].unique():
    class_df = combined_df[combined_df['Label'] == label]
    cap = SAMPLES_PER_CLASS.get(label, 150000)
    if len(class_df) > cap:
        class_df = class_df.sample(n=cap, random_state=42)
    sampled_dfs.append(class_df)
    print(f"{label}: {len(class_df):,} rows")

combined_df = pd.concat(sampled_dfs, ignore_index=True)
print(f"\nFinal sampled shape: {combined_df.shape}")
print(combined_df['Label'].value_counts())
```

```python
print(f"Dataset memory usage: {combined_df.memory_usage(deep=True).sum() / 1e9:.2f} GB")
```

```python
columns_to_drop = ['Flow ID', 'Src IP', 'Dst IP', 'Timestamp']
combined_df = combined_df.drop(columns=columns_to_drop)
print(f"Shape after dropping ID columns: {combined_df.shape}")
```

```python
before = len(combined_df)
combined_df = combined_df.drop_duplicates()
after = len(combined_df)
print(f"Removed {before - after:,} duplicate rows")
print(f"Shape after removing duplicates: {combined_df.shape}")
```

```python
import numpy as np

# Replace infinities with NaN
numeric_cols = combined_df.select_dtypes(include=[np.number]).columns
combined_df[numeric_cols] = combined_df[numeric_cols].replace([np.inf, -np.inf], np.nan)

# Check how many missing values we now have
print(f"Missing values per column (top 10):")
print(combined_df.isnull().sum().sort_values(ascending=False).head(10))
print(f"\nTotal missing values: {combined_df.isnull().sum().sum()}")
```

```python
before = len(combined_df)
combined_df = combined_df.dropna()
after = len(combined_df)
print(f"Removed {before - after:,} rows with missing values")
print(f"Shape after dropping missing values: {combined_df.shape}")
```

```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()
combined_df['Label_encoded'] = label_encoder.fit_transform(combined_df['Label'])

# See the mapping so we know which number = which class
label_mapping = dict(zip(label_encoder.classes_, label_encoder.transform(label_encoder.classes_)))
print("Label mapping:")
for label, code in label_mapping.items():
    print(f"  {label} → {code}")

print(f"\nShape: {combined_df.shape}")
```

```python
from sklearn.feature_selection import SelectKBest, f_classif

# Separate features and target
X = combined_df.drop(columns=['Label', 'Label_encoded'])
y = combined_df['Label_encoded']

# Check for any remaining non-numeric columns that would break ANOVA
non_numeric = X.select_dtypes(exclude=[np.number]).columns.tolist()
print(f"Non-numeric columns remaining: {non_numeric}")
```

```python
selector = SelectKBest(score_func=f_classif, k=30)
X_selected = selector.fit_transform(X, y)

# Get the names of the selected features
selected_features = X.columns[selector.get_support()].tolist()
print(f"Selected {len(selected_features)} features:")
for f in selected_features:
    print(f"  - {f}")
```

```python
constant_features = X.columns[[33, 34, 35, 50, 57, 58, 59]].tolist()
print("Constant features (zero variance):")
for f in constant_features:
    print(f"  - {f}")
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_selected, y, test_size=0.2, stratify=y, random_state=42
)

print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"y_test shape: {y_test.shape}")

print(f"\nTrain class distribution:")
print(y_train.value_counts(normalize=True))
print(f"\nTest class distribution:")
print(y_test.value_counts(normalize=True))
```

```python
import pandas as pd

X_train_df = pd.DataFrame(X_train, columns=selected_features)
X_test_df = pd.DataFrame(X_test, columns=selected_features)
y_train_df = pd.DataFrame(y_train, columns=['Label_encoded']).reset_index(drop=True)
y_test_df = pd.DataFrame(y_test, columns=['Label_encoded']).reset_index(drop=True)

print(f"X_train_df: {X_train_df.shape}")
print(f"X_test_df: {X_test_df.shape}")
print(f"y_train_df: {y_train_df.shape}")
print(f"y_test_df: {y_test_df.shape}")
```
